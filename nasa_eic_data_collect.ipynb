```python
import requests

search_url = "https://svs.gsfc.nasa.gov/api/search/"
search_response = requests.get(
    search_url,
    params={"keywords": "Earth Information Center", "limit": 100},
    timeout=30,
)
search_response.raise_for_status()
results = search_response.json().get("results") or []

video_uris = []
for entry in results:
    page_id = entry["id"]
    page_response = requests.get(
        f"https://svs.gsfc.nasa.gov/api/{page_id}/",
        timeout=30,
    )
    page_response.raise_for_status()
    page_data = page_response.json()

    for group in page_data.get("media_groups") or []:
        for item in group.get("items") or []:
            if not isinstance(item, dict):
                continue
            instance = item.get("instance") or {}
            asset_url = instance.get("url") or ""
            if asset_url.endswith((".mp4", ".webm")):
                video_uris.append(asset_url)

for uri in video_uris:
    print(uri)
```
